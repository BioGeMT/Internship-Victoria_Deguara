```python
import pandas as pd
from miRBench.dataset import get_dataset_df 

df1=get_dataset_df('AGO2_CLASH_Hejret2023', split="test")[['gene', 'noncodingRNA', 'label']]
df2=get_dataset_df('AGO2_eCLIP_Klimentova2022', split="test")[['gene', 'noncodingRNA', 'label']]
df3=get_dataset_df("AGO2_eCLIP_Manakov2022", split='test')[['gene', 'noncodingRNA', 'label']]
df4=get_dataset_df("AGO2_eCLIP_Manakov2022", split='leftout')[['gene', 'noncodingRNA', 'label']]
datasets= {"AGO2_CLASH_Hejret2023": df1,
    "AGO2_eCLIP_Klimentova2022": df2,
    "AGO2_eCLIP_Manakov2022_test":df3,
    "AGO2_eCLIP_Manakov2022_leftout":df4
    }

def get_seed6mer(noncodingRNA):
    return [noncodingRNA[1:7],noncodingRNA[2:8],'A'+noncodingRNA[1:6]]

def get_seed7mer(noncodingRNA) :
    return [noncodingRNA[1:8],'A'+noncodingRNA[1:7]]

def get_seed8mer(noncodingRNA):
    return ['A'+noncodingRNA[1:8]]

for name, df in datasets.items():
    df['seed6mer (1:7)(2:8)(A+1:6)']=df['noncodingRNA'].apply(get_seed6mer)
    print (f'----{name}----')
    display(df[['noncodingRNA','seed6mer (1:7)(2:8)(A+1:6)']].head())

for name, df in datasets.items():
    df['seed7mer (1:8)(A+1:7)']=df['noncodingRNA'].apply(get_seed7mer)
    print (f'----{name}----')
    display(df[['noncodingRNA','seed7mer (1:8)(A+1:7)']].head())

for name, df in datasets.items():
    df['seed8mer (A+1:8)']=df['noncodingRNA'].apply(get_seed8mer)
    print (f'----{name}----')
    display(df[['noncodingRNA', 'seed8mer (A+1:8)']].head())

df1.to_csv("AGO2_CLASH_Hejret2023.csv", index=False)
df2.to_csv("AGO2_eCLIP_Klimentova2022.csv", index=False)
df3.to_csv("AGO2_eCLIP_Manakov2022_test.csv", index=False)
df4.to_csv("AGO2_eCLIP_Manakov2022_leftout.csv", index=False)
```
